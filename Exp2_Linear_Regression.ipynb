```python
# here to implement simple linear regression will perform some steps
# Step 1 : importing required librabries
# Step 2 : read the data
# Step 3 : Model selection and fitting
# Step 4 : predicting the value using linear equation y=mx+c
# Step 5 : visualizing the data
```

**Step 1 :Importing Library**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

**Step 2 : Read the Data**

```python
data = {"hours" : [1,2,3,4,5],
        "Marks" : [50,60,80,90,100]}
df = pd.DataFrame(data)
print(df)
```

```python
x = df[["hours"]].values
y = df[["Marks"]].values
```

**Model Fitting**

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x,y)
```

```python
slope = model.coef_[0]
intercept = model.intercept_
```

```python
# "f-string" (formatted string literal) in python
# .2f : This is a format specifier used within the curly braces of an f-string
print(f"slope(m):{slope}")
print(f"Intercept (b):{intercept}")
print(f"Equation of the regreesion line: y={slope[0]:.2f} x + {intercept[0]:.2f}")
```

```python
df['Predicted Marks'] = model.predict(x)
```

```python
# here keep in mind the data which we have provided to the model is also linear
df.head()
```

**Visualizing**

```python
plt.scatter(df['hours'],df['Marks'],color ='blue',label='Actual Marks')
```

**plotting the equation**

```python
plt.plot(df['hours'],df['Predicted Marks'],color='red',label='Regression Line')
plt.xlabel('Hours studied')
plt.ylabel('Marks')
plt.title('Linear Regression:Hours vs Marks')
plt.legend()
plt.show()
```
