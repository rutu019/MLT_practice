**Step 1 : Data Collection**

Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('/content/drive/MyDrive/play_tennis.csv')
```

```python
df.head()
```

**Step 2 :Data Preprocessing**

converting data into suitable format for model creation

Encoding

```python
le=LabelEncoder()
df['outlook_n']=le.fit_transform(df['outlook'])
df['temp_n']=le.fit_transform(df['temp'])
df['humidity_n']=le.fit_transform(df['humidity'])
df['wind_n']=le.fit_transform(df['wind'])
df['play_n']=le.fit_transform(df['play'])
```

```python
df
```

```python
df=df.drop(['outlook','temp','humidity','wind','play'],axis='columns')
```

```python
df
```

**Step 3 : Spliting:Input and Output variable i.e Independent and Dependent variables**

```python
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
```

```python
independent_variable=df.drop(['play_n','day'],axis='columns')#X
independent_variable = independent_variable.astype(float)
dependent_variable=df['play_n']#Y
```

```python
model=tree.DecisionTreeClassifier()
# it is used to create an instance and decision tree classifier model in scikit-learn
```

```python
model.fit(independent_variable,dependent_variable)
# fit method in machine learning is used to train a model on given dataset
```

```python
model.score(independent_variable,dependent_variable)
```

```python
model.predict([[1,2,0,1]])
```

**Testing of Our model(Checking the accuracy)**

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
# Split data into 70% train and 30% test
X_train, X_test, Y_train, Y_test = train_test_split(independent_variable, dependent_variable, test_size=0.3, random_state=42)
```

```python
# Train the model on training data
model = tree.DecisionTreeClassifier()
model.fit(X_train, Y_train)
```

```python
# predict on test data
Y_pred  = model.predict(X_test)
```

```python
# Evaluate accuracy
accuracy = accuracy_score(Y_test, Y_pred)
print(f"Accuracy on test data: {accuracy:.2f}")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Detailed classification report
print("\nClassification Report:")
print(classification_report(Y_test, Y_pred))
```

```python
# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(Y_test, Y_pred))
```
